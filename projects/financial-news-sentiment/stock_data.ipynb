```python
import yfinance as yf
import pandas as pd

# Define ticker and date range
ticker_symbol = 'AAPL'
start_date = '2015-01-01'
end_date = '2024-12-31'

# Download stock data
stock_data = yf.download(ticker_symbol, start=start_date, end=end_date)

# Ensure the index is a DatetimeIndex
stock_data.index = pd.to_datetime(stock_data.index)

# Create a full date range from start to end
date_range = pd.date_range(start=start_date, end=end_date, freq='D')

# Reindex stock_data to include all dates, filling missing values with forward-fill
stock_data = stock_data.reindex(date_range, method='ffill')

# Rename index for clarity
stock_data.index.name = 'Date'

# Add a flag to mark rows that were originally missing
stock_data['is_filled'] = ~stock_data.index.isin(stock_data.dropna().index)

# Save to CSV
stock_data.to_csv('data/Apple_stock_data_filled.csv', index=True)

# Print preview
print(stock_data.head())
```

```python
import pandas as pd

df = pd.read_csv("data/Apple_stock_data_filled.csv", skiprows=[1], index_col=0)  # Skip second row & set index

# Drop the duplicate first row if it exists
df = df[df.index != "Date"]

# Convert index to datetime if needed
df.index = pd.to_datetime(df.index, errors='coerce')

# Ensure numeric columns are actually numbers
df = df.apply(pd.to_numeric, errors='coerce')

df.index.name = "Date"  # Rename the index from 'Price' to 'Date'

print(df.head())  # Check if correctly formatted

df.to_csv('data/Apple_stock_data_filled.csv', index=True)
```
