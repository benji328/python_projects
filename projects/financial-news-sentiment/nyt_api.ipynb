```python
import os
import pandas as pd
import time
import requests

API_KEY = os.getenv("NYT_API_KEY")
if not API_KEY:
    raise RuntimeError("Set NYT_API_KEY in your environment before running this notebook.")
BASE_URL = "https://api.nytimes.com/svc/search/v2/articlesearch.json"

def fetch_articles(query, begin_date, end_date, page=0):
    params = {
        "q": query,
        "begin_date": begin_date,
        "end_date": end_date,
        "api-key": API_KEY,
        "page": page
    }
    
    response = requests.get(BASE_URL, params=params)
    
    if response.status_code == 200:
        data = response.json()
        articles = data.get("response", {}).get("docs", [])
        return [
            {
                "headline": article.get("headline", {}).get("main", ""), 
                "pub_date": article.get("pub_date", "")
            }
            for article in articles
        ]
    else:
        print(f"Error: {response.status_code} - {response.text}")
        return []

# Modify this section to query for Apple
query = "Apple"
begin_date = "20150101"  # January 1, 2015
end_date = "20161231"    # December 31, 2016
all_articles = []


pg = 1000
p = 0
for page in range(0, pg):
    try:
        articles = fetch_articles(query, begin_date, end_date, page)
        if not articles:
            break  
        all_articles.extend(articles)
        time.sleep(3)
    except:
        p=page
        print(p)

df = pd.DataFrame(all_articles)
print(df)

df.to_csv("data/1516apple_headlines.csv", index=False)
```

```python
for page in range(pg, pg+100):
    articles = fetch_articles(query, begin_date, end_date, page)
        
    if not articles:
        break  
        
    all_articles.extend(articles)
    time.sleep(12)

df = pd.DataFrame(all_articles)
print(df)
df.to_csv("data/tesla_headlines.csv", index=False)
```
